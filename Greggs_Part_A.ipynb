## 1. Part A — Advice on the decision

### The decision

Greggs must decide whether to consolidate in-house manufacturing, as proposed on 30 September 2026 [S1 pp.2–3; S2]. The proposal closes Enfield, North Lakes, Pettigrews and Seaham, ends manufacturing at Treforest (retained for distribution), narrows ranges at Clydesmill and Manchester, ends tinned bread at Gosforth and buys a few products from specialist suppliers. Six manufacturing sites would remain. The executive team under CEO Roisin Currie proposes it; we assume the board decides after consultation. We found no final decision by 25 November, although consultation was reported to run to the end of November [S20]. Changes would start no earlier than Q2 2027 [S10].

The alternatives, and the capital spending that differs between them (Exhibit 2), are:

1. **Proceed.** Cash cost about £60m, including capex of about £40m [S1 p.3]. We assume the capex covers all relocated lines, including any at Derby: £24m in FY2027 and £16m in FY2028.
2. **Scale back.** Close fewer sites or lines: lower spending, on the same timing, and lower savings.
3. **Keep the network** and add growth capacity in reserved space at Derby (room for five production platforms, with further frozen lines planned [S6 pp.28–29]) and Kettering [S6 pp.31–32]. The company discloses no cost or firm timetable for this.

Derby (£135m cash budget, including its first line) and Kettering (£105m after land) [S5 p.17] proceed under all three, as does ordinary capital spending (guidance: £150–170m a year from 2027 [S4 p.15]).

### Relevant costs and benefits

Only future cash flows that differ matter (Exhibit 1). Proceeding adds capex, redundancy and disruption payments. It saves wages, utilities, maintenance and site costs that stop, less the extra cost of making or buying the products elsewhere; we assume the company's saving is net of this.

Capacity and opportunity costs also count. Derby space taken by relocated lines is unavailable for later growth. Receiving sites must absorb volume without weakening resilience: losing Balliol Park's savoury supply is already a stress scenario [S3 p.69]. Sale or lease value from vacated sites would be a benefit, assumed nil; the 2024 Twickenham sale raised £14.9m [S3 p.148] but says little about these sites.

Under alternative 3, products stay where they are now made (Exhibit 1): for example rolls and pizza at Enfield and Manchester, sandwich relishes and pizza sauce at Pettigrews, and bread at Gosforth and Treforest [S6 p.18]. Growth goes to Derby and Kettering. Buying a product instead avoids its ingredients, direct labour and any manufacturing overhead that actually stops [S3 p.143]; whole-site fixed costs stop only if the site closes. Buy only if the supplier's price plus transition cost is below that avoidable cost plus the value of freed space. Greggs' own test is to make only where bought-in cost and quality can be beaten [S6 p.10]. No supplier prices are disclosed.

We exclude the more than £300m invested in manufacturing and logistics since 2024 [S2], other past spending, carrying values, depreciation and write-offs (sunk or non-cash, apart from tax or sale effects), and head-office allocations that continue. The new distribution centres' 2027 overheads [S1 p.3] arise under every alternative.

### Estimates

The company expects an annual pre-tax cash saving of about £20m from completion, realised across FY2028 and FY2029 [S1 p.3]. Some press reports read as £20m over two years [S10]; the company's wording is annual. We assume half arrives in FY2028 and all from FY2029, flat to FY2036, and spread the £20m non-capex cost 30%, 50% and 20% over FY2027–29. We assume Treforest, where press reports about 101 roles at risk [S11], is inside the £60m, £20m and 740 roles; the update does not say.

The saving is 7.9% of 2025 supply costs of £253.4m [S3 p.145]: plausible, not proven. Precedent helps only partly. After the 2016 programme, 2019 net supply cost was £9.5m below 2015, against £7.0m expected [S18 p.21], while expected one-off cash exceptional costs rose from £25m to £27m [S17 p.31; S18 p.19]. That programme also planned £100m of investment, partly for distribution [S6 p.11], so it does not validate £20m from £60m.

Site costs are undisclosed, and cost behaviour cannot be estimated reliably because the supply-cost line gives no site or volume detail. Regressing half-year supply costs on shops trading (four points) gives £0.083m per shop and a negative fixed cost (Exhibit 6); a rough inflation restatement more than halves the slope. This is not a unit variable cost: shops are not production volume, and the slope absorbs inflation and capacity steps, so we do not use it. The accounts themselves warn that internal supply-chain cost cannot be cut quickly [S3 p.69].

### The analysis

We discount year-end cash flows to 1 January 2027 at 9.5%, the pre-tax cost of capital used for impairment testing [S3 p.135]. It was set for shop assets, so we test 8–15%.

Against alternative 3, proceeding has a **net present value of £46.8m** (costs £52.1m, savings £99.0m), an internal rate of return of 30%, and cumulative cash turns positive in 2031 (Exhibit 3). After 25% tax [S12], assuming immediate relief on all costs, it is £43.8m. Deducting the 10% profit share from the pre-tax case [S4 p.9] leaves £38.6m.

Taken one at a time, value reaches zero if the annual saving falls to £10.5m (47% below forecast) or cash cost reaches £113.9m. It turns negative if savings slip four years or stop before 2032, or if proceeding needs more than £51.3m of undisclosed extra capex in 2027 (Exhibit 4). Reading £20m as a two-year total would give −£36.2m. Combined, a £15m saving, 25% overspend, one-year delay and cash flows ending in 2033 give −£22.8m: joint slippage over a short life destroys the case.

Alternative 2, illustratively keeping 60% of savings for 70% of cost, is worth £22.9m; at that cost it needs 84% of savings to match proceeding. In general, each closure should earn annual savings above about 18p per £1 of its cash cost (Exhibit 5).

Alternative 3 is the zero baseline. Every £10m of capacity spending it would need, but proceeding avoids, adds £9.1m to proceeding's advantage.

### The recommendation

**Proceed, conditionally**, approving closures in stages after consultation:

1. Each closure or supplier switch clears the 18p test on avoidable cash costs; drop those that fail (alternative 2).
2. A validated saving of at least £15m, on schedule, and cash cost no higher than £75m, which still gives £9.1m. If the programme cannot meet this, rework it before committing capital.
3. Receiving capacity is proven before each closure, with a tested fallback for disruption.
4. Supplier contracts price below avoidable cost and protect quality and second sourcing.
5. Consultation alternatives, including the union's [S16], are costed on the same basis.

Management still needs site-level avoidable costs, output, headcount and utilisation; yearly phasing; capex by site, including Derby; tenure, disposal value and exit costs; supplier quotes; Treforest's treatment; and alternative 3's capacity spending.

### Make-or-buy logic (alternative 3 and the specialist-supplier part of alternative 1)

Buy a product if: **supplier price × volume + transition cost < avoidable make cost + value of the capacity freed**

| Cost of making in-house (S3 p.143 classification) | Avoided by buying? |
|---|---|
| Direct materials (ingredients, packaging) | Yes |
| Direct labour | Yes, once roles actually end (redundancy is a one-off cost) |
| Manufacturing overhead: utilities, maintenance | Only the part that stops; site fixed costs only if the site closes |
| Depreciation of existing plant | No: non-cash and sunk |
| Allocated central overhead | No |

No volumes or prices are disclosed, so no per-unit comparison is possible.

## 6. Timeline (workbook tab) — cash by year per alternative — **Exhibit 2**

All flows are **incremental to alternative 3**. Negative = cash out.

```python
def flows(**kw):
    p = {**BASE, **kw}
    last = p["end"]
    rows = []
    for y in range(2027, last + 1):
        ys = y - p["delay"]
        share = 1.0 if ys >= p["full"] else p["ramp"].get(ys, 0.0)
        sav = p["saving"] * share * (1 + p["growth"]) ** max(0, y - 2029)
        capex = -(p["capex"] * p["capex_ph"].get(y, 0) + (p["extra_capex"] if y == 2027 else 0))
        oth = -(p["other"] * p["other_ph"].get(y, 0) + (p["extra_other"] if y == 2027 else 0))
        disp = p["disposal"] if y == 2029 else 0.0
        maint = p["maint"] if y >= 2029 else 0.0
        ps = -V["profit_share"]/100 * p["profit_share"] * (sav + oth)
        t = p["tax"]
        taxcf = -t * (sav + oth + ps + (capex if p["capex_relief"] else 0.0))
        rows.append((y, capex + 0.0, oth + 0.0, sav, disp + maint, ps + taxcf + 0.0))
    return pd.DataFrame(rows, columns=["Year","Capex","Disruption & redundancy","Operating saving","Disposal / avoided capex","Profit share & tax"]).set_index("Year")

def npv(**kw):
    p = {**BASE, **kw}
    cf = flows(**kw).sum(axis=1)
    r = p["r"] * (1 - p["tax"])
    t = cf.index.values - 2026 - (0.5 if p["mid"] else 0.0)
    return float((cf.values / (1 + r) ** t).sum())

a1 = flows()
a2 = flows(saving=BASE["saving"]*A2_f, capex=BASE["capex"]*A2_g, other=BASE["other"]*A2_g)
a3 = a1 * 0
disc = pd.Series(1 / (1 + BASE["r"]) ** (a1.index - 2026), index=a1.index)

assert np.isclose(-a1["Capex"].sum(), V["capex"]) and np.isclose(-a1["Disruption & redundancy"].sum(), V["cash_cost"]-V["capex"])

tl = pd.DataFrame({
    "Alt 1: capex": a1["Capex"], "Alt 1: disruption & redundancy": a1["Disruption & redundancy"],
    "Alt 1: saving": a1["Operating saving"], "ALT 1 NET": a1.sum(axis=1),
    "Alt 2: capex": a2["Capex"], "Alt 2: disruption & redundancy": a2["Disruption & redundancy"],
    "Alt 2: saving": a2["Operating saving"], "ALT 2 NET": a2.sum(axis=1),
    "ALT 3 NET (baseline)": a3.sum(axis=1),
    "Discount factor": disc, "Alt 1 PV": a1.sum(axis=1) * disc, "Alt 2 PV": a2.sum(axis=1) * disc,
    "Alt 1 cumulative cash": a1.sum(axis=1).cumsum(),
}).T
tl["Total"] = tl.sum(axis=1)
tl.loc[["Discount factor","Alt 1 cumulative cash"], "Total"] = np.nan
display(Markdown("**Exhibit 2. Cash by year per alternative, £m, incremental to alternative 3** (phasing and alternative 2 shares are assumptions A2–A4, A14)"))
display(tl.style.format("{:,.1f}", na_rep="").format("{:.3f}", subset=pd.IndexSlice[["Discount factor"], :], na_rep=""))

memo = pd.DataFrame([
 ("Programme capex","c.£40m, FY2027–28 (A2)","Part of £40m (illustration £28m)","None"),
 ("Relocated lines, incl. at Derby","Assumed inside £40m (A9); amount not disclosed","Fewer lines","None"),
 ("Growth capacity at Derby / Kettering","Still needed for growth beyond relocated volume; not disclosed","Not disclosed","Needed sooner or larger on this route: not disclosed (A13)"),
 ("Derby £135m / Kettering £105m + £30m land","Committed or spent","Same","Same"),
 ("Contracted commitments £84.0m (Dec 2025, 2026–27)","Committed","Same","Same"),
 ("Company guidance: £180m in 2026; £150–170m a year from 2027","£40m is about an eighth of two years' guidance; the filings read do not say whether it is inside it","Guidance set before the proposal","Guidance set before the proposal"),
], columns=["Capital spending","Alternative 1","Alternative 2","Alternative 3"]).set_index("Capital spending")
display(Markdown("**Exhibit 2 (continued). Which capital spending, under which alternative**"))
table(memo)
```

## 7. NPV and sensitivity (workbook tab) — **Exhibits 3–5**

```python
def solve(f, lo, hi, tol=1e-9):
    flo = f(lo)
    assert flo * f(hi) <= 0, "root not bracketed"
    while hi - lo > tol:
        m = (lo + hi) / 2
        if f(m) * flo > 0: lo, flo = m, f(m)
        else: hi = m
    return (lo + hi) / 2

R = {}
R["npv_a1"] = npv()
R["pv_cost"] = -npv(saving=0)
R["pv_sav"] = R["npv_a1"] + R["pv_cost"]
R["irr"] = solve(lambda r: npv(r=r), 0.01, 1.0) * 100
cum = a1.sum(axis=1).cumsum(); R["payback"] = int(cum[cum >= 0].index[0])
cumd = (a1.sum(axis=1) * disc).cumsum(); R["dpayback"] = int(cumd[cumd >= 0].index[0])
R["npv_tax"] = npv(tax=V["tax"]/100)
R["npv_ps"] = npv(profit_share=1.0)
R["npv_tax_nocapex"] = npv(tax=V["tax"]/100, capex_relief=False)
R["npv_a2"] = npv(saving=BASE["saving"]*A2_f, capex=BASE["capex"]*A2_g, other=BASE["other"]*A2_g)
R["be_saving"] = solve(lambda s: npv(saving=s), 0, 40)
R["be_saving_fall"] = (1 - R["be_saving"] / V["saving"]) * 100
R["be_cost_mult"] = solve(lambda m: npv(capex=BASE["capex"]*m, other=BASE["other"]*m), 0.5, 5)
R["be_cost"] = R["be_cost_mult"] * V["cash_cost"]
R["be_delay"] = next(d for d in range(0, 10) if npv(delay=d) < 0)
R["be_end"] = next(e for e in range(2028, 2042) if npv(end=e) > 0)
R["be_extra"] = solve(lambda x: npv(extra_capex=x), 0, 200)
R["per_pound"] = R["be_saving"] / V["cash_cost"] * 100          # pence of annual saving per £1 of cash cost
R["a2_need_f"] = (R["npv_a1"] + A2_g * R["pv_cost"]) / R["pv_sav"] * 100
R["alt3_per10"] = 10 / (1 + BASE["r"])      # PV of £10m spent in FY2027
R["press"] = npv(saving=10, ramp={2028: 1.0}, full=2029, end=2029)  # £10m in 2028 and 2029 only
R["combined"] = npv(saving=15, capex=BASE["capex"]*1.25, other=BASE["other"]*1.25, delay=1, end=2033)
R["floor_saving"], R["floor_cost"] = 15, 75
R["floor"] = npv(saving=15, capex=BASE["capex"]*75/60, other=BASE["other"]*75/60)
R["prog_over"] = (V["prog_cost_2019"] / V["prog_cost_2016"] - 1) * 100
R["prog_beat"] = (V["prog_saving_2019"] / V["prog_saving_plan"] - 1) * 100
R["npv_prog_over"] = npv(capex=BASE["capex"]*(1+R["prog_over"]/100), other=BASE["other"]*(1+R["prog_over"]/100))
R["prog_yield"] = V["prog_saving_2019"] / (V["prog_capex"] + V["prog_cost_2019"]) * 100   # pence per £1, not like for like
R["pct_supply"] = V["saving"] / V["supply_fy25"] * 100
R["pct_rev"] = V["saving"] / V["revenue_fy25"] * 100
R["per_role"] = V["saving"] / V["roles"] * 1000
R["capex_share"] = V["capex"] / (V["guide_lo"] + V["guide_hi"]) * 100

# alternative 2 threshold, direct check: NPV(a2 at required share) equals NPV(a1)
f_req = R["a2_need_f"]/100
assert np.isclose(npv(saving=20*f_req, capex=40*A2_g, other=20*A2_g), R["npv_a1"])

summary = pd.DataFrame([
 ("1. Proceed", R["npv_a1"], R["pv_cost"], R["pv_sav"]),
 ("2. Scale back (illustration 60% savings, 70% cost)", R["npv_a2"], R["pv_cost"]*A2_g, R["pv_sav"]*A2_f),
 ("3. Keep network (baseline, no extra capex)", 0.0, 0.0, 0.0),
], columns=["Alternative","NPV","PV of costs","PV of savings"]).set_index("Alternative")
display(Markdown(f"**Exhibit 3. Net present value at {V['rate']}% pre-tax, £m, 1 Jan 2027**"))
display(summary.style.format("{:,.1f}"))
display(Markdown(f"IRR {R['irr']:.1f}% · cumulative cash positive in {R['payback']} · discounted payback {R['dpayback']} · "
                 f"after-tax NPV {R['npv_tax']:.1f} (full relief) or {R['npv_tax_nocapex']:.1f} (no relief on capex)"))

fig, ax = plt.subplots(1, 2, figsize=(11, 3.6))
net = a1.sum(axis=1)
ax[0].bar(net.index, net.values, color=np.where(net.values < 0, RED, BLUE), width=0.7)
ax[0].axhline(0, color=INK, lw=0.8); ax[0].set_title("Alternative 1: net cash flow by year, £m"); ax[0].set_xticks(net.index, [str(y)[2:] for y in net.index]); ax[0].set_xlabel("financial year (20..)")
for label, d, col in [("1. Proceed", a1, BLUE), ("2. Scale back (illustration)", a2, ORANGE)]:
    s = (d.sum(axis=1) * disc).cumsum()
    ax[1].plot(s.index, s.values, color=col, lw=2, marker="o", ms=4, label=label)
    ax[1].annotate(f"{s.iloc[-1]:.1f}", (s.index[-1], s.iloc[-1]), xytext=(6, 0), textcoords="offset points", va="center", color=INK)
ax[1].axhline(0, color=MUTED, lw=1, ls="--", label="3. Keep network (baseline)")
ax[1].set_title("Cumulative present value, £m"); ax[1].set_xticks(net.index, [str(y)[2:] for y in net.index]); ax[1].legend(frameon=False)
plt.tight_layout(); plt.show()
```

## 8. Cost behaviour (workbook tab) — **Exhibit 6**

Data: segment *supply costs* are reported for **FY2024, FY2025, H1 2025 and H1 2026** only in the filings read (H1 2025 results do not show the line for H1 2024). H2 2025 is derived as FY2025 − H1 2025. To compare like with like we use half-year figures; FY2024 enters as a half-year average.

```python
cb = pd.DataFrame([
 ("FY2024 (half-year average)", V["shops_fy24"], V["supply_fy24"]/2, V["revenue_fy24"]/2),
 ("H1 2025", V["shops_h1_25"], V["supply_h1_25"], V["revenue_h1_25"]),
 ("H2 2025 (derived)", V["shops_fy25"], V["supply_fy25"]-V["supply_h1_25"], V["revenue_fy25"]-V["revenue_h1_25"]),
 ("H1 2026", V["shops_h1_26"], V["supply_h1_26"], V["revenue_h1_26"]),
], columns=["Period","Shops at period end","Supply costs £m","Revenue £m"]).set_index("Period")
cb["% of revenue"] = cb["Supply costs £m"] / cb["Revenue £m"] * 100

x, y = cb["Shops at period end"].values.astype(float), cb["Supply costs £m"].values.astype(float)
def ols(x, y):
    b, a = np.polyfit(x, y, 1)
    r2 = 1 - ((y - (a + b*x))**2).sum() / ((y - y.mean())**2).sum()
    return a, b, r2

def hilo(x, y):
    i, j = x.argmin(), x.argmax()
    b = (y[j] - y[i]) / (x[j] - x[i])
    return y[i] - b*x[i], b

a, b, r2 = ols(x, y); ha, hb = hilo(x, y)
# restate at H2 2025 prices using disclosed like-for-like inflation (rough: 5.5% 2025, 2.2% H1 2026)
y_real = y * np.array([1.055, 1.0, 1.0, 1/1.022])
ra, rb, rr2 = ols(x, y_real)
keep = [0, 1, 3]; da, db, dr2 = ols(x[keep], y[keep])
R["slope"], R["intercept"], R["r2"], R["slope_real"] = b, a, r2, rb

est = pd.DataFrame([
 ("OLS regression, 4 half-year points", a, b, r2),
 ("High-low (fewest v. most shops)", ha, hb, np.nan),
 ("OLS, roughly restated for inflation", ra, rb, rr2),
 ("OLS, dropping the derived H2 2025 point", da, db, dr2),
], columns=["Method","Intercept £m per half-year","Slope £m per extra shop","R²"]).set_index("Method")

display(Markdown("**Exhibit 6. Supply costs against shops trading — shown to demonstrate why it is not used**"))
display(cb.style.format({"Shops at period end":"{:,.0f}","Supply costs £m":"{:.1f}","Revenue £m":"{:,.1f}","% of revenue":"{:.1f}"}))
display(est.style.format({"Intercept £m per half-year":"{:.1f}","Slope £m per extra shop":"{:.3f}","R²":"{:.2f}"}, na_rep="–"))

fig, ax = plt.subplots(figsize=(8, 3.8))
ax.scatter(x, y, s=55, color=BLUE, zorder=3, label="Reported / derived half-years")
xx = np.linspace(x.min()-10, x.max()+10, 50)
ax.plot(xx, a + b*xx, color=BLUE, lw=2, label=f"OLS: {a:.1f} + {b:.3f} × shops (R² {r2:.2f})")
ax.plot(xx, ha + hb*xx, color=ORANGE, lw=2, ls="--", label=f"High-low: {ha:.1f} + {hb:.3f} × shops")
for name, xi, yi in zip(cb.index, x, y):
    ax.annotate(name.replace(" (half-year average)"," avg").replace(" (derived)","*"), (xi, yi), xytext=(5,-10), textcoords="offset points", fontsize=8, color=MUTED)
ax.set_xlabel("shops trading at period end (axis does not start at zero)"); ax.set_ylabel("supply costs, £m per half-year"); ax.legend(frameon=False, fontsize=8.5)
plt.tight_layout(); plt.show()
```

**What the regression says about the decision: essentially nothing.**

1. The driver is a count of shops, not a volume of production, so the slope is **not a unit variable cost**.
2. A negative intercept would mean negative fixed cost, which is impossible. The line only describes a narrow range far from zero, and a rough inflation restatement turns the intercept positive and more than halves the slope: the estimate is unstable.
3. Four points (one derived, one an average) leave two degrees of freedom; a high R² can arise from any two series that both trend upward.
4. The network changed in the period (new capacity, Derby preparation), and prices inflated.
5. The segment line has no breakdown between manufacturing and logistics, or by site, in the pages read, so it cannot show what stops when a site closes.

The course methods are therefore shown but **not used**; the company's headline saving is used as a labelled assumption (A5) and tested by thresholds.
